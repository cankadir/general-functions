```python
import pandas as pd
import requests
path = "https://api.airtable.com/v0/appu8ocaGPbj3Vt2k/indicators?view=indicatorSorting"
readOnlyKey = 'pattDYj5xTUxYtdsI.d8be15f625755afe3659c505731a777c3eebd5fc9f2ab822a567e24a47e824b7'

headers =  {
          'Content-Type': 'application/json',
          'Authorization': f'Bearer {readOnlyKey}'
          }

r = requests.get(path, headers=headers )
data = r.json()

data = data['records']
data
```

```python
import time

for i,r in df_img.set_index('Indicator').iterrows():
    
    print( r['Topic'] , i)
    # Get the images, URL's
    if not pd.isna(r['bg-img']): bg_img = r['bg-img'][0]['url']
    
    callout_imgs = []
    if isinstance(r['Callout_img'], list):
        # Handle the case where Callout_img is a list
        for callout_img in r['Callout_img']:
            callout_imgs.append(callout_img['url'])
    elif pd.isna(r['Callout_img']):
        continue
    else:
        # Handle other cases if necessary
        continue

    charts = []
    for viz in ['Viz-1','Viz-2','Viz-3', 'Viz-4', 'Viz-5', 'Viz-6', 'Viz-7']:
        if not pd.isna(r[viz]):
            charts.append(r[viz][0]['url'])

    # Download
    # 1. save BG Image
    response = requests.get(bg_img)

    if response.status_code == 200:
        with open( os.path.join(target_folder, r['Topic'],i, f"{i}_bg_img.jpg"), "wb") as file:
            file.write(response.content)

    # 2. save Callout Images
    for idx, callout_img in enumerate(callout_imgs):
        response = requests.get(callout_img)

        if response.status_code == 200:
            with open( os.path.join(target_folder, r['Topic'],i, f"{i}_callout_img_{idx}.jpg"), "wb") as file:
                file.write(response.content)

    # 3. save Charts
    for idx, chart in enumerate(charts):
        response = requests.get(chart)

        if response.status_code == 200:
            with open( os.path.join(target_folder, r['Topic'],i, f"{i}_chart_{idx}.jpg"), "wb") as file:
                file.write(response.content)

    time.sleep( 2 )
```
